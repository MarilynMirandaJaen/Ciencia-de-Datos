```python
import pandas as pd
import matplotlib.pyplot as plt

datos = {
    "Meses": [
        "Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio",
        "Julio", "Agosto", "Setiembre", "Octubre", "Noviembre", "Diciembre"
    ],
    "Tienda física (%)": [
        84.00, 82.67, 81.90, 79.92, 78.27, 75.93,
        74.43, 72.37, 70.88, 68.60, 67.19, 64.29
    ],
    "En línea (%)": [
        16.00, 17.33, 18.10, 20.08, 21.73, 24.07,
        25.57, 27.63, 29.12, 31.40, 32.81, 35.71
    ]
}

df = pd.DataFrame(datos)

ax = df.plot(
    x="Meses",
    y=["Tienda física (%)", "En línea (%)"],
    kind="bar",
    figsize=(14,6),
    width=0.8,
    color=["green", "skyblue"]
)

plt.title("Porcentaje de ventas por tienda")
plt.xlabel("Mes")
plt.ylabel("Porcentaje (%)")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Tipo de tienda")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

trimestres = ["Q1", "Q2", "Q3", "Q4"]

anio1 = [3, 2, 2, 9]
anio2 = [4, 3, 2, 10]
anio3 = [3, 2, 3, 11]

plt.figure(figsize=(9, 5))

plt.plot(trimestres, anio1, marker="o", color="blue", label="Año 1")
plt.plot(trimestres, anio2, marker="o", color="green", label="Año 2")
plt.plot(trimestres, anio3, marker="o", color="red", label="Año 3")

plt.title("Renuncias por trimestre")
plt.xlabel("Trimestre")
plt.ylabel("Cantidad de renuncias")

plt.legend()
plt.grid(True)

plt.show()
```
