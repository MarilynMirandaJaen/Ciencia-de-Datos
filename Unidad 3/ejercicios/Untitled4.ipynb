```python
import pandas as pd

datos = {
    "Comercio": ["A", "B", "C", "D", "E", "F", "G"],
    "Visitas": [2, 4, 1, 6, 3, 5, 2],
    "Pedido": [45, 68, 30, 95, 52, 80, 40]
}

df = pd.DataFrame(datos)

print(df)

# Calcular correlación de Pearson
correlacion = df["Visitas"].corr(df["Pedido"])

print("\nCoeficiente de correlación:", correlacion)
```

```python
import pandas as pd

comercios = []
visitas = []
pedidos = []

cantidad = int(input("Ingrese la cantidad de comercios: "))

for i in range(cantidad):
    print("\nComercio", i + 1)

    comercio = input("Ingrese el nombre del comercio: ")
    visita = int(input("Ingrese la cantidad de visitas: "))
    pedido = float(input("Ingrese el monto del pedido: "))

    comercios.append(comercio)
    visitas.append(visita)
    pedidos.append(pedido)

datos = {
    "Comercio": comercios,
    "Visitas": visitas,
    "Pedido": pedidos
}

df = pd.DataFrame(datos)

print("\nDatos ingresados:")
print(df)

# Calcular correlación de Pearson
correlacion = df["Visitas"].corr(df["Pedido"])

print("\nCoeficiente de correlación:", correlacion)
```

```python
import matplotlib.pyplot as plt

datos = [38, 42, 44, 45, 47, 48, 49, 50, 52, 53, 55, 58, 60, 310]

limite_inferior = 27.5
limite_superior = 71.5

plt.figure(figsize=(10, 5))

# Gráfico de barras
plt.bar(range(1, len(datos) + 1), datos)

# Límites
plt.axhline(limite_inferior, color="green",
            linestyle="--", label="Límite inferior = 27.5")

plt.axhline(limite_superior, color="red",
            linestyle="--", label="Límite superior = 71.5")

plt.title("Detección de valores atípicos")
plt.xlabel("Posición del dato")
plt.ylabel("Valor")

plt.xticks(range(1, len(datos) + 1))
plt.legend()

plt.show()
```
