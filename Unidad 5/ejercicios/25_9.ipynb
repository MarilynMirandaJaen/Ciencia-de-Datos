```python
# 1. Importar librerías

import numpy as np
import pandas as pd

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 2. Crear 300 vuelos

np.random.seed(8)

n = 300

hora = np.random.randint(0, 24, n)          # hora de salida
clima = np.random.randint(0, 2, n)          # 0 = bueno, 1 = malo
trafico = np.random.randint(0, 3, n)        # 0 = bajo, 1 = medio, 2 = alto
distancia = np.random.randint(100, 3000, n) # distancia en km

# Crear retrasos
riesgo = clima * 2 + trafico + (hora >= 18)

# Agregamos un poco de azar para que los datos no sean perfectos
riesgo = riesgo + np.random.normal(0, 1.5, n)

retraso = np.where(riesgo >= 2.5, "Sí", "No")

# Crear tabla

vuelos = pd.DataFrame({
    "hora": hora,
    "clima": clima,
    "trafico": trafico,
    "distancia": distancia,
    "retraso": retraso
})

print(vuelos.head())

# 3. Separar X y Y

X = vuelos[["hora", "clima", "trafico", "distancia"]]

y = vuelos["retraso"]


# Dividir 80% entrenamiento y 20% prueba

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Entrenamiento:", len(X_train))
print("Prueba:", len(X_test))

# 4. Crear y entrenar modelo simple

modelo_simple = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

modelo_simple.fit(X_train, y_train)

# 5. Crear y entrenar modelo complejo

modelo_complejo = DecisionTreeClassifier(
    random_state=42
)

modelo_complejo.fit(X_train, y_train)

# 6. Comparar los modelos

print("\nMODELO SIMPLE")

print("Entrenamiento:",
      round(accuracy_score(
          y_train,
          modelo_simple.predict(X_train)
      ) * 100, 2), "%")

print("Prueba:",
      round(accuracy_score(
          y_test,
          modelo_simple.predict(X_test)
      ) * 100, 2), "%")


print("\nMODELO COMPLEJO")

print("Entrenamiento:",
      round(accuracy_score(
          y_train,
          modelo_complejo.predict(X_train)
      ) * 100, 2), "%")

print("Prueba:",
      round(accuracy_score(
          y_test,
          modelo_complejo.predict(X_test)
      ) * 100, 2), "%")

# 7. Predecir un vuelo nuevo

vuelo = pd.DataFrame({
    "hora": [20],
    "clima": [1],
    "trafico": [2],
    "distancia": [1500]
})

prediccion = modelo_simple.predict(vuelo)[0]

print("\n¿Se retrasará el vuelo?", prediccion)
```

```python
# 1. Importar librerías

import pandas as pd

from sklearn.datasets import load_wine
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 2. Cargar datos reales de vinos

vino = load_wine()

datos = pd.DataFrame(
    vino.data,
    columns=vino.feature_names
)

datos["tipo"] = vino.target

print(datos.head())
print("\nCantidad de vinos:", len(datos))

# 3. Separar X y Y

X = datos[vino.feature_names]    # características del vino

y = datos["tipo"]                # tipo de vino

# 4. Dividir entrenamiento y prueba

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Entrenamiento:", len(X_train))
print("Prueba:", len(X_test))

# 5. Crear y entrenar modelo

modelo = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

modelo.fit(X_train, y_train)

# 6. Evaluar modelo

prediccion = modelo.predict(X_test)

exactitud = accuracy_score(y_test, prediccion)

print(
    "Exactitud del modelo:",
    round(exactitud * 100, 2), "%"
)

# 7. Predecir un vino

vino_nuevo = X_test.iloc[[0]]

resultado = modelo.predict(vino_nuevo)[0]

print("\nTipo de vino predicho:", vino.target_names[resultado])
```
