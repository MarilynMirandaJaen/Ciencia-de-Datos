```python
#sklearn es una biblioteca de Python especializada en Machine Learning.
from sklearn.tree import DecisionTreeClassifier

#datos
X = [
    [18, 1],
    [20, 2],
    [25, 5],
    [30, 7],
    [35, 9]
]

# Respuestas
y = [
    "No",
    "No",
    "Sí",
    "Sí",
    "Sí"
]
# Crear modelo
modelo = DecisionTreeClassifier()

# Entrenar fit() significa: "Aprende con estos datos."
modelo.fit(X, y)

# Nuevo cliente: Aquí se crea un nuevo cliente.
nuevo_cliente = [[27, 6]]
# Predicción con modelo.predict le pedimos a Python que de una predicción.
prediccion = modelo.predict(nuevo_cliente)

print("La Predicción es: ", prediccion[0])
```
