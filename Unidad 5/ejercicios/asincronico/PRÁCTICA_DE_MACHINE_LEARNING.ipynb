**1. ¿Es regresión o clasificación? ¿Por qué?**

Es un problema de clasificación porque queremos predecir si un socio va a cancelar o no su membresía. Solo existen dos posibles resultados: Sí o No.

**2. Identifica X y Y.**

X son las variables que utilizaremos para hacer la predicción:

* meses_activo
* visitas_mes
* quejas
* plan

Y es la variable que queremos predecir:

* cancelo

**3. ¿Sería buena idea utilizar “recibió encuesta de salida” como X?**

No sería buena idea porque esa información se obtiene después de que el socio cancela. El modelo estaría utilizando información del futuro para realizar la predicción.


```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import confusion_matrix
```

```python
from google.colab import files

uploaded = files.upload()

df = pd.read_csv("gimnasio_cancelaciones(1).csv")

df.head()
```

```python
print(df["cancelo"].value_counts())
```

```python
df["plan_mensual"] = df["plan"].map({
    "Mensual": 1,
    "Anual": 0
})

df["cancelo"] = df["cancelo"].map({
    "Sí": 1,
    "No": 0
})

df.head()
```

```python
X = df[["meses_activo", "visitas_mes", "quejas", "plan_mensual"]]

y = df["cancelo"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    stratify=y,
    random_state=42
)
```

```python
modelo_logistico = LogisticRegression()

modelo_logistico.fit(X_train, y_train)

pred_logistico = modelo_logistico.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, pred_logistico))
print("Precision:", precision_score(y_test, pred_logistico))
print("Recall:", recall_score(y_test, pred_logistico))
print("Matriz de confusión:")
print(confusion_matrix(y_test, pred_logistico))
```

```python
arbol = DecisionTreeClassifier(random_state=42)

arbol.fit(X_train, y_train)

pred_arbol = arbol.predict(X_test)

print("Accuracy:", accuracy_score(y_test, pred_arbol))
print("Precision:", precision_score(y_test, pred_arbol))
print("Recall:", recall_score(y_test, pred_arbol))
print("Matriz de confusión:")
print(confusion_matrix(y_test, pred_arbol))
```

```python
arbol_limitado = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

arbol_limitado.fit(X_train, y_train)

pred_limitado = arbol_limitado.predict(X_test)

print("Accuracy:", accuracy_score(y_test, pred_limitado))
print("Precision:", precision_score(y_test, pred_limitado))
print("Recall:", recall_score(y_test, pred_limitado))
print("Matriz de confusión:")
print(confusion_matrix(y_test, pred_limitado))
```

En este ejercicio se probaron tres modelos para predecir si un socio cancelará su membresía. La regresión logística obtuvo el mejor resultado en los datos de prueba, con aproximadamente un 93 % de accuracy. El árbol sin límite obtuvo un resultado menor y puede ser más propenso al sobreajuste. Por estos resultados, para este conjunto de datos usaría la regresión logística.


```python
print("REGRESIÓN LOGÍSTICA")
print("Entrenamiento:", modelo_logistico.score(X_train, y_train))
print("Prueba:", modelo_logistico.score(X_test, y_test))

print("\nÁRBOL SIN LÍMITE")
print("Entrenamiento:", arbol.score(X_train, y_train))
print("Prueba:", arbol.score(X_test, y_test))

print("\nÁRBOL LIMITADO")
print("Entrenamiento:", arbol_limitado.score(X_train, y_train))
print("Prueba:", arbol_limitado.score(X_test, y_test))
```

```python
from sklearn.model_selection import cross_val_score
```

```python
cv_logistico = cross_val_score(
    modelo_logistico, X, y, cv=5
)

cv_arbol = cross_val_score(
    arbol, X, y, cv=5
)

cv_limitado = cross_val_score(
    arbol_limitado, X, y, cv=5
)

print("Regresión logística:", cv_logistico.mean())
print("Árbol sin límite:", cv_arbol.mean())
print("Árbol limitado:", cv_limitado.mean())
```

La validación cruzada sirve para evaluar el modelo varias veces utilizando diferentes partes de los datos. En este caso se utilizaron 5 grupos. Al final se obtiene el promedio de los resultados y así podemos tener una mejor idea de cómo funciona cada modelo con datos diferentes.
