```python
#Variables independientes y dependientes
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from google.colab import files
import io
#Creamos la tabla con 4 telefonos ya vendidos
telefonos = pd.DataFrame ({
    "Meses_uso":  [3, 18, 30, 8], #Antiguedad del telefono
    "Bateria_pct":[95,82,70,90], #Procentaje de bateria
    "Precio":     [420,260,150,380]  #Precio en dolares
})
telefonos #Mostrar la tabla par revisar los datos
print(telefonos)
# X son las pistas: las columnas que el modelo usa para predecir
X = telefonos [["Meses_uso","Bateria_pct"]]  #dos columnas -> va en doble corchete
# y es la repuesta: la columna que quereos calcular
y = telefonos ["Precio"]

# Creamos y entrenado el modelo con las pistas a la vez

modelo_telefonos = LinearRegression()
modelo_telefonos.fit(X,y)
print("Modelo entrenado")
#Variables independientes y dependientes

#Probamos el modelo con un telefono nuevo 12 meses u 85% de bateria
telefono_nuevo = pd.DataFrame({"Meses_uso":[12],"Bateria_pct":[85]}) #mismos nombres de columna
precio_estimado = modelo_telefonos.predict(telefono_nuevo)           #el modelo calcula el precio modelo

print(f"Precio estimado: ¢{precio_estimado[0]:.0f} mil, compre mejor el HONOR nuevo")     # mostramos el resultado redondeado, sin decimales
```
